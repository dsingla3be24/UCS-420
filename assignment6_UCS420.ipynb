### Q1

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original temperature:", temperature)

temp_corrected = temperature + 2
print("\na. Corrected temperatures (+2°C):", temp_corrected)

temp_fahrenheit = (9/5) * temp_corrected + 32
print("\nb. Temperatures in Fahrenheit:", temp_fahrenheit)

greater_than_32 = temperature[temperature > 32]
print("\nc. Readings greater than 32°C:", greater_than_32)

count_exceed_32 = np.sum(temperature > 32)
print("\nd. Number of readings exceeding 32°C:", count_exceed_32)
```

**e. Ans(Vectorization):**
Vectorization and Boolean indexing make data processing more efficient because they push the loop operations down into the highly optimized C-level code underlying NumPy, avoiding the overhead of Python's for loops. This allows for parallelized, contiguous memory access operations which execute significantly faster.

### Q2

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

total_steps = np.sum(steps)
print("a. Total steps recorded:", total_steps)

mean_steps = np.mean(steps)
print("\nb. Mean number of steps:", mean_steps)

max_steps = np.max(steps)
min_steps = np.min(steps)
print("\nc. Maximum steps:", max_steps, "| Minimum steps:", min_steps)

steps_per_day = np.sum(steps, axis=0)
print("\nd. Total steps for each day:", steps_per_day)

steps_per_user = np.sum(steps, axis=1)
print("\ne. Total steps for each user:", steps_per_user)

max_index_flat = np.argmax(steps)
user_day_position = np.unravel_index(max_index_flat, steps.shape)
print(f"\nf. Maximum steps are at position (User, Day): {user_day_position}")
```

### Q4

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print("a. Shape of X:", X.shape)
print("   Dimensions of X (ndim):", X.ndim)

X_T = X.T
print("\nb. Transpose of X (X.T):\n", X_T)
```

**Ans(Explanation for Transpose):**
The transpose X.T flips the matrix over its diagonal. Originally, rows were users and columns were features. In the transposed matrix, the rows are the features (Sleep hours, Activity level, Stress level) and the columns represent the individual users.

```python
XTX = X.T @ X
print("c. Product of X.T and X (X.T @ X):\n", XTX)

XTX_inv = np.linalg.inv(XTX)
print("\nd. Inverse of (X.T @ X):\n", XTX_inv)

beta = XTX_inv @ X.T @ y
print("\ne. Calculated coefficients (beta):", beta)
```

**f. Ans(Explanation of coefficients:)**
The resulting coefficients ($\beta$) represent the weights or importance of each feature in predicting the assistance score. 
- $\beta_0$ corresponds to the weight for 'Sleep hours'
- $\beta_1$ corresponds to the weight for 'Activity level'
- $\beta_2$ corresponds to the weight for 'Stress level'

For every one-unit increase in a feature, the assistance score changes by the value of its corresponding coefficient, assuming all other features remain constant.

```python
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta
print("g. Predicted assistance score for the new user:", predicted_score)
```
