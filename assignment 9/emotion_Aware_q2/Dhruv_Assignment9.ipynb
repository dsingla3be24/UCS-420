```python
from PIL import Image
from landingai.predict import Predictor
from gtts import gTTS
from IPython.display import Audio, display

# 1. Credentials
endpoint_id = "c7b0f834-0a9c-4f52-9e8f-96f401cc621d"
api_key = "land_sk_zpzoFc0VMmmQ2lIW0kUxR1ta6YjhsQTzViYiSvhBGhzHwNgybl"

# 2. Image load
image_path = "test.jpg"  # You can also test with happy.jpg or angry.jpg
image = Image.open(image_path)

# 3. Prediction
predictor = Predictor(endpoint_id, api_key=api_key)
predictions = predictor.predict(image)

print("Output:", predictions)

# 4. Result message
if len(predictions) > 0:
    label = predictions[0].label_name
    confidence = round(predictions[0].score * 100, 1)
    voice_msg = f"You are looking {label} with {confidence} percent confidence."
else:
    voice_msg = "No emotion detected."

print("Result:", voice_msg)

# 5. Audio play
tts = gTTS(text=voice_msg, lang='en')
tts.save("voice_output.mp3")
display(Audio("voice_output.mp3", autoplay=True))
```
