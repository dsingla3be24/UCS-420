### Question 3: Random Numbers & Seed

#### i. Generate list of 100 random numbers between 100 and 900 (inclusive)

```python
import random
from collections import Counter

random.seed(1024170434)
randomnumber = [random.randint(100, 900) for i in range(100)]
print(randomnumber)
```

#### ii. Count and print all odd numbers in the list

```python
oddnumber = [x for x in randomnumber if x % 2 != 0]
print("Odd numbers count:", len(oddnumber))
print("Odd numbers:", oddnumber)
```

#### iii. Count and print all even numbers in the list

```python
evennumber = [x for x in randomnumber if x % 2 == 0]
print("Even numbers count:", len(evennumber))
print("Even numbers:", evennumber)
```

#### iv. Count and print all prime numbers, and build list of prime numbers

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

primenumber = [x for x in randomnumber if is_prime(x)]
print("Prime numbers count:", len(primenumber))
print("Prime numbers:", primenumber)
```

#### v. Print the number that occurs most frequently and how many times it occurs

```python
counts = Counter(randomnumber)
most_common_num, freq = counts.most_common(1)[0]
print("Most freq number:", most_common_num, "Freq:", freq)
```
